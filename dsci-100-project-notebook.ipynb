#### Task 5.2 Perform the Analysis (20 points)

- Perform the proposed analysis.
- Be sure to comment your code throughout so that it is easy to follow. 

*You may add more code cells if needed.* 


</div>

```r
library(tidymodels)


set.seed(42)
march_split<-initial_split(march_tidy, prop=0.75)
march_train<-training(march_split)
march_test<- testing(march_split)

lm_spec<-linear_reg()|>
set_engine('lm')|>
set_mode('regression')

lm_recipe<-recipe(conf_wins ~ reg_wins + conf_wins_pct, data = march_train)

lm_fit<-workflow()|>
add_recipe(lm_recipe)|>
add_model(lm_spec)|>
fit(data = march_train)

lm_test_results<-lm_fit|>
predict(new_data=march_test)|>
bind_cols(march_test)|>
metrics(truth = conf_wins, estimate = .pred)

lm_rmspe <- lm_test_results|>
filter(.metric == 'rmse')|>
select(.estimate)|>
pull()

lm_rmspe


```

This visualization displays a scatter plot comparing our model's predicted conference wins against the actual conference wins (conf_wins) for the unseen test dataset. The red reference line indicates an ideal 1:1 relationship where predictions perfectly match actual outcomes. By observing how closely the data points cluster around this line, we can visually assess the accuracy and bias of our multiple linear regression model. the points above the line represent underestimations, while points below represent overestimations. This plot is essential for interpreting our quantitative RMSPE metric, as it reveals whether our errors are evenly distributed across teams or if the model struggles with specific win ranges.


```r
plot_data<-lm_fit|>
predict(new_data=march_test)|>
bind_cols(march_test)

viz_final<-ggplot(plot_data, aes(x= reg_wins, y = conf_wins))+
geom_point(color = 'dark green', alpha= 0.7, size=3)+
geom_line(aes(y = .pred), color='red')+
labs(title = 'Actual vs. Predicted Conference Wins', x = 'Regular Season Wins', y = 'Conference Wins')+
theme(text = element_text(size = 14))

viz_final

```
